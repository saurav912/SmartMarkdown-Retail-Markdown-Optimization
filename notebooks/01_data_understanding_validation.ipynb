#### Feature Availability and Leakage Audit

##### Purpose:
This step documents how selected dataset variables should be treated before
building the demand prediction model.

For each important feature, we check:
1. Whether the variable would realistically be available at prediction time.
2. Whether it should be used directly, only historically/lagged, or avoided.
3. Whether using it could introduce data leakage.
4. Whether the variable has any special limitation, such as Latent_Demand
   being available only because the dataset is synthetic.

##### Prediction setup:
We assume information available up to Week t will be used to predict demand
for Week t+1.

This audit helps ensure that future information is not accidentally used
during model training.


```python
feature_audit = pd.DataFrame({
    "Column": [
        "Units_Sold",
        "Closing_Inventory",
        "Views",
        "Cart_Adds",
        "Revenue",
        "Gross_Profit",
        "Promotion_Flag",
        "Retail_Event",
        "Latent_Demand"
    ],

    "Use_in_Model": [
        "Historical only",
        "Historical/current state",
        "Historical or lagged",
        "Historical or lagged",
        "Avoid as direct predictor",
        "Avoid as direct predictor",
        "Only if known in advance",
        "Yes - calendar known",
        "Special synthetic variable"
    ],

    "Risk": [
        "Low",
        "Low",
        "Medium",
        "Medium",
        "High leakage risk",
        "High leakage risk",
        "Depends on timing",
        "Low",
        "Real-world availability issue"
    ]
})

display(feature_audit)
```

```python
# Selling price should not exceed MRP

price_above_mrp = df_raw[df_raw["Selling_Price"] > df_raw["MRP"]]

print("Selling_Price > MRP:", len(price_above_mrp))


# Selling price should normally remain above Unit Cost
price_below_cost = df_raw[df_raw["Selling_Price"] <= df_raw["Unit_Cost"]]

print("Selling_Price <= Unit_Cost:", len(price_below_cost))
```

```python
# Discount validation

expected_discount_pct = (
    (df_raw["MRP"] - df_raw["Selling_Price"])
    / df_raw["MRP"]
    * 100
)

discount_difference = (
    expected_discount_pct - df_raw["Discount_Pct"]
).abs()

print(
    "Discount_Pct mismatches:",
    (discount_difference > 0.01).sum()
)

print("Minimum Discount_Pct:", df_raw["Discount_Pct"].min())
print("Maximum Discount_Pct:", df_raw["Discount_Pct"].max())
print("Maximum discount calculation difference:", discount_difference.max())
```

##### Available Inventory = Opening Inventory + Replenishment Quantity

```python
# ---------------------------------------------------------
# 3. Inventory reconciliation
# ---------------------------------------------------------

expected_available_inventory = (
    df_raw["Opening_Inventory"]
    + df_raw["Replenishment_Qty"]
)

print(
    "Available_Inventory mismatches:",
    (
        df_raw["Available_Inventory"]
        != expected_available_inventory
    ).sum()
)


expected_closing_inventory = (
    df_raw["Available_Inventory"]
    - df_raw["Units_Sold"]
    + df_raw["Returns_Qty"]
)

print(
    "Closing_Inventory mismatches:",
    (
        df_raw["Closing_Inventory"]
        != expected_closing_inventory
    ).sum()
)
```

##### Sales cannot exceed available stock

```python
# Sales vs available inventory

print(
    "Units_Sold > Available_Inventory:",
    (
        df_raw["Units_Sold"]
        > df_raw["Available_Inventory"]
    ).sum()
)
```

##### Checking for negative or impossible values

```python
non_negative_columns = [
    "MRP",
    "Unit_Cost",
    "Selling_Price",
    "Opening_Inventory",
    "Replenishment_Qty",
    "Available_Inventory",
    "Latent_Demand",
    "Units_Sold",
    "Returns_Qty",
    "Closing_Inventory",
    "Days_In_Inventory",
    "Competitor_Price",
    "Views",
    "Cart_Adds",
    "Revenue",
    "Gross_Profit"
]

negative_check = pd.DataFrame({
    "Column": non_negative_columns,
    "Negative_Count": [
        (df_raw[col] < 0).sum()
        for col in non_negative_columns
    ]
})

display(negative_check)
```

##### Revenue Validation

Revenue = Selling_Price × Units_Sold

```python
# Revenue validation

expected_revenue = (
    df_raw["Selling_Price"]
    * df_raw["Units_Sold"]
)

revenue_difference = (
    expected_revenue - df_raw["Revenue"]
).abs()

print(
    "Revenue mismatches:",
    (revenue_difference > 0.01).sum()
)

print(
    "Maximum Revenue difference:",
    revenue_difference.max()
)
```

##### Gross Profit Validation

Gross Profit = (Selling Price - Unit Cost) × Units Sold

```python
expected_gross_profit = ((df_raw["Selling_Price"] - df_raw["Unit_Cost"])* df_raw["Units_Sold"])

gross_profit_difference = (expected_gross_profit - df_raw["Gross_Profit"]).abs()

print("Gross Profit mismatches:",(gross_profit_difference > 0.01).sum())

print("Maximum Gross Profit difference:",gross_profit_difference.max())
```

##### Marging Percentage Validation

Margin % = (Selling Price - Unit Cost) / Selling Price × 100

```python
expected_margin_pct = (
    (df_raw["Selling_Price"] - df_raw["Unit_Cost"])
    / df_raw["Selling_Price"]
    * 100
)

margin_difference = (
    expected_margin_pct
    - df_raw["Margin_Pct"]
).abs()

print(
    "Margin_Pct mismatches:",
    (margin_difference > 0.01).sum()
)

print(
    "Maximum Margin difference:",
    margin_difference.max()
)
```

```python
# Stockout and latent-demand consistency

# Case 1:
# Latent demand exceeds available inventory.
# These rows should normally be marked as stockouts.

demand_exceeds_inventory = (
    df_raw["Latent_Demand"]
    > df_raw["Available_Inventory"]
)

print(
    "Rows where Latent_Demand > Available_Inventory:",
    demand_exceeds_inventory.sum()
)
```

```python
# Of the above rows, check whether Stockout_Flag = 1

stockout_flag_mismatch_1 = df_raw[
    (df_raw["Latent_Demand"] > df_raw["Available_Inventory"])
    &
    (df_raw["Stockout_Flag"] != 1)
]

print(
    "Demand > Inventory but Stockout_Flag != 1:",
    len(stockout_flag_mismatch_1)
)
```

```python
# If Stockout_Flag = 1, latent demand should generally
# exceed available inventory.

stockout_flag_mismatch_2 = df_raw[
    (df_raw["Stockout_Flag"] == 1)
    &
    (df_raw["Latent_Demand"] <= df_raw["Available_Inventory"])
]

print(
    "Stockout_Flag = 1 but Demand <= Inventory:",
    len(stockout_flag_mismatch_2)
)
```

```python
stockout_sales_mismatch = df_raw[
    (df_raw["Latent_Demand"] > df_raw["Available_Inventory"])
    &
    (df_raw["Units_Sold"] != df_raw["Available_Inventory"])
]

print(
    "Stockout rows where Units_Sold != Available_Inventory:",
    len(stockout_sales_mismatch)
)
```

```python
# Stockout frequency

stockout_summary = df_raw["Stockout_Flag"].value_counts()

print(stockout_summary)

print(
    "\nStockout Rate:",
    round(df_raw["Stockout_Flag"].mean() * 100, 2),
    "%"
)
```

```python
# Difference between true demand and observed sales

df_check = df_raw.copy()

df_check["Unmet_Demand"] = (
    df_check["Latent_Demand"]
    - df_check["Units_Sold"]
)

print(
    "Rows with unmet demand:",
    (df_check["Unmet_Demand"] > 0).sum()
)

print(
    "Average unmet demand when positive:",
    round(
        df_check.loc[
            df_check["Unmet_Demand"] > 0,
            "Unmet_Demand"
        ].mean(),
        2
    )
)

print(
    "Maximum unmet demand:",
    df_check["Unmet_Demand"].max()
)
```

## Stage 1 - Validation Summary

- Dataset contains 40 SKUs with 105 continuous weekly observations per SKU.
- Total observations: 4,200.
- Date range: 01-Jan-2024 to 29-Dec-2025.
- No duplicate SKU-week records or exact duplicate rows were found.
- Retail_Event nulls represent non-event weeks and are therefore structurally valid.
- Pricing, discount, inventory, revenue, gross profit and margin relationships were validated successfully.
- No invalid negative values were found in the checked business variables.
- 112 observations (2.67%) represent stockout periods.
- All stockout observations are internally consistent:
  Latent_Demand > Available_Inventory and Units_Sold is capped by available stock.
- Average unmet demand during stockout periods is 31.62 units, with a maximum of 92 units.
- The distinction between Units_Sold and Latent_Demand will be considered during EDA before finalizing the demand-regression target.

Conclusion:
The dataset is structurally and logically validated and is ready for exploratory data analysis.
